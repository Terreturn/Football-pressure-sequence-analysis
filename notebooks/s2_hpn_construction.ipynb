# S2 — HPN construction and visualisation

The notebook randomly selects valid, informative freeze frames and sequences from the user-generated S1 labels. No match or event identifier is required.


```python
from pathlib import Path
import os, sys, pandas as pd
ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(ROOT))
EVENTS_DIR, F360_DIR, OUTPUT_DIR = Path(os.environ['EVENTS_DIR']), Path(os.environ['F360_DIR']), Path(os.environ['OUTPUT_DIR'])
labels = pd.read_csv(OUTPUT_DIR / 'labels.csv')
from src.hpn_network import PressureParams
PRESSURE_PARAMS = PressureParams(player_distance=6.4, player_sd=2.2, boundary_distance=3.2, boundary_sd=1.1)
RANDOM_STATE = 202405
```

## One randomly selected network: layer-by-layer construction


```python
from src.hpn_network import sample_network_frame, plot_network_layers

network = sample_network_frame(labels, EVENTS_DIR, F360_DIR, random_state=RANDOM_STATE, params=PRESSURE_PARAMS)
fig = plot_network_layers(network)
fig.show()
```

```python
from src.hpn_network import sample_network_sequence, plot_network_sequence

networks = sample_network_sequence(labels, EVENTS_DIR, F360_DIR, max_frames=5, random_state=RANDOM_STATE, params=PRESSURE_PARAMS)
fig = plot_network_sequence(networks)
fig.show()
```
