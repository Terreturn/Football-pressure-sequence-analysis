# S1 — Detect and label high-press sequences

Set the three environment variables described in the README, then run this notebook.


```python
from pathlib import Path
import os, sys
ROOT = Path.cwd().resolve().parent if Path.cwd().name == 'notebooks' else Path.cwd().resolve()
sys.path.insert(0, str(ROOT))
EVENTS_DIR = Path(os.environ['EVENTS_DIR'])
F360_DIR = Path(os.environ['F360_DIR'])
OUTPUT_DIR = Path(os.environ['OUTPUT_DIR'])
print(EVENTS_DIR, F360_DIR, OUTPUT_DIR, sep='\n')
```

```python
from src.data_pipeline import LabelConfig, SequenceConfig, build_sequences, save_stage1
from src.hpn_network import PressureParams

# Final-main defaults. Edit these objects to analyse a different pressure or label definition.
PRESSURE_PARAMS = PressureParams(player_distance=6.4, player_sd=2.2, boundary_distance=3.2, boundary_sd=1.1)
SEQUENCE_CONFIG = SequenceConfig(pressure_threshold=.65, own_half_x=60.0, max_gap_seconds=5.0)
LABEL_CONFIG = LabelConfig()

sequences, labels = build_sequences(
    EVENTS_DIR, F360_DIR, SEQUENCE_CONFIG, params=PRESSURE_PARAMS, label_config=LABEL_CONFIG,
)
sequence_path, label_path = save_stage1(sequences, labels, OUTPUT_DIR)
display(labels['outcome_tag'].value_counts(dropna=False).to_frame('anchors'))
print('Saved:', sequence_path, label_path)
```
